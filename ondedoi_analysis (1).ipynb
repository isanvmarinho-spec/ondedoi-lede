#Standardizing the data in a column that aggregates specialties regardless of spelling or number of specialties listed.
#1 Decide if you want to count responses or mentions (some include two)
#2 Give the list to Chat GPT with every variation and ask it to create a clean list and a dictionary
#3 Make a note that some people included two specialties in the same response, so I will count mentions instead of responses. Explain that in the article.

```python
df = df.rename(columns={
    "9 - Qual era a especialidade do profissional?": "perp_specialty"
})

df["response_id"] = df.index #give each row a number

def extract_specialties(value):  #this is a function that turns each messy answer into a list of specialties
    if pd.isna(value):
        return []

    text = value.strip().lower()

    specialties = []

    if "gine" in text or "genicologista" in text or "obstetr" in text or text == "go":
        specialties.append("Gynecology/obstetrics")

    if "pediatra" in text:
        specialties.append("Pediatrics")

    if "clinico" in text or "clínico" in text or "clínica" in text or "clinica" in text or "geral" == text:
        specialties.append("General practice/internal medicine")

    if "ortoped" in text or "traumato" in text:
        specialties.append("Orthopedics/traumatology")

    if "cardio" in text:
        specialties.append("Cardiology")

    if "dentista" in text or "odontolog" in text:
        specialties.append("Dentistry")

    if "cirurgi" in text or "cirurgião" in text:
        specialties.append("Surgery")

    if "ultra" in text or "raio x" in text or "ecografo" in text:
        specialties.append("Radiology")

    if "psiquiatr" in text or "psicolog" in text or "psicana" in text:
        specialties.append("Psychiatry/psychology")

    if "nutri" in text or "nutro" in text:
        specialties.append("Nutrition")

    if "enferm" in text:
        specialties.append("Nursing")

    if "dermatolog" in text:
        specialties.append("Dermatology")

    if "otorrino" in text:
        specialties.append("Otorhinolaryngology")

    if "gastro" in text:
        specialties.append("Gastroenterology")

    if "homeopatia" in text:
        specialties.append("Homeopathy")

    if "fisioterapia" in text or "fisioterapeuta" in text:
        specialties.append("Physiotherapy")

    if "reumato" in text:
        specialties.append("Rheumatology")

    if "anestesi" in text:
        specialties.append("Anesthesiology")

    if "mastologista" in text:
        specialties.append("Mastology")

    if len(specialties) == 0:
        specialties.append("Other/unclear")

    return specialties
```

```python
df["specialty_list"] = df["perp_specialty"].apply(extract_specialties)

#explode it into a specialty-level dataset

df_specialty = df.explode("specialty_list")

df_specialty = df_specialty.rename(columns={
    "specialty_list": "specialty"
})

specialty_counts = df_specialty["specialty"].value_counts()

specialty_percentages = (
    df_specialty["specialty"].value_counts(normalize=True) * 100
).round(1)

print(specialty_counts)
print(specialty_percentages)
```

```python
# Question 1: Are people of different economic status more vulnerable to abuse by a specific type of health professional or location?
# This shows patterns only, not proof of risk.

#ANSWER
# The general idea is:
# 1. Within each economic-status group, what percentage of cases happened in private practice, private hospital, free clinic, free hospital, etc.?
# 2. Within each economic-status group, what percentage of cases happened in each specialty?

econ_location_grouped = df.groupby(["victim_econ_status", "offense_location_clean"])  # group by class and place
econ_by_location = econ_location_grouped.size()  # count rows in each group
econ_by_location = econ_by_location.reset_index(name="count")  # turn counts into a table

location_totals = econ_by_location.groupby("victim_econ_status")["count"].transform("sum")  # total for each class

econ_by_location["percentage_within_economic_status"] = econ_by_location["count"] / location_totals  # make a share
econ_by_location["percentage_within_economic_status"] = econ_by_location["percentage_within_economic_status"] * 100  # make it percent
econ_by_location["percentage_within_economic_status"] = econ_by_location["percentage_within_economic_status"].round(1)  # round it 

print(econ_by_location) 

top_specialties = ["Gynecology/obstetrics", "Pediatrics", "General practice/internal medicine"]  # pick top groups

econ_specialty_grouped = df_specialty.groupby(["victim_econ_status", "specialty"])  # group by class and specialty
econ_by_specialty = econ_specialty_grouped.size()  # count rows in each group. this one returns something that looks like a table, but technically the group names are stored in the index, not regular columns.
econ_by_specialty = econ_by_specialty.reset_index(name="count")  # turns the grouped result back into a normal table, and names the count column "count"

specialty_totals = econ_by_specialty.groupby("victim_econ_status")["count"].transform("sum")  # total for each class

econ_by_specialty["percentage_within_economic_status"] = econ_by_specialty["count"] / specialty_totals  # make a share
econ_by_specialty["percentage_within_economic_status"] = econ_by_specialty["percentage_within_economic_status"] * 100  # make it percent
econ_by_specialty["percentage_within_economic_status"] = econ_by_specialty["percentage_within_economic_status"].round(1)  # round it 

econ_by_top_specialties = econ_by_specialty[econ_by_specialty["specialty"].isin(top_specialties)]  # keep top groups

print(econ_by_top_specialties)  
```

```python
# Question 2: What type of abuse is more common among the top 3 specialties?
# This counts abuse mentions, not just respondents.

top_specialties = ["Gynecology/obstetrics", "Pediatrics", "General practice/internal medicine"]  # pick top groups

top_specialty_abuse = df_specialty[df_specialty["specialty"].isin(top_specialties)].copy()  # keep top groups
top_specialty_abuse["abuse_type"] = top_specialty_abuse["types_abuse"].str.split(",")  # split multiple answers
top_specialty_abuse = top_specialty_abuse.explode("abuse_type")  # one abuse type per row
top_specialty_abuse["abuse_type"] = top_specialty_abuse["abuse_type"].str.strip()  # clean spaces

abuse_specialty_grouped = top_specialty_abuse.groupby(["specialty", "abuse_type"])  # group by specialty and abuse
abuse_by_specialty = abuse_specialty_grouped.size()  # count rows in each group
abuse_by_specialty = abuse_by_specialty.reset_index(name="count")  # turn counts into a table

abuse_totals = abuse_by_specialty.groupby("specialty")["count"].transform("sum")  # total for each specialty

abuse_by_specialty["percentage_within_specialty"] = abuse_by_specialty["count"] / abuse_totals  # make a share
abuse_by_specialty["percentage_within_specialty"] = abuse_by_specialty["percentage_within_specialty"] * 100  
abuse_by_specialty["percentage_within_specialty"] = abuse_by_specialty["percentage_within_specialty"].round(1) 

abuse_by_specialty = abuse_by_specialty.sort_values(["specialty", "count"], ascending=[True, False])  # sort biggest first

print("Abuse type by top specialty")  # label the table
print(abuse_by_specialty)  # show the table
```

```python
abuse_by_specialty.to_csv("abuse_by_specialty.csv")
```

```python
# Question 3: Are underage victims (< 18) more likely to be abused by a specific type of health professional?
# The 16-20 category is excluded because it mixes minors and adults.

df["age_group"] = "Age range overlaps 18"  # start with unclear ages
df.loc[df["victim_age"].isin(["2-5", "6-10", "11-15", "15-17"]), "age_group"] = "Under 18"  # mark minors
df.loc[df["victim_age"].isin(["18-30", "20-30", "31-40", "41-50", "50+"]), "age_group"] = "18+"  # mark adults

df_age_specialty = df.explode("specialty_list")  # one specialty per row
df_age_specialty = df_age_specialty.rename(columns={"specialty_list": "specialty"})  # rename the column
df_age_specialty = df_age_specialty[df_age_specialty["age_group"].isin(["Under 18", "18+"])]  # keep clear ages

age_specialty_grouped = df_age_specialty.groupby(["age_group", "specialty"])  # group by age and specialty
specialty_by_age = age_specialty_grouped.size()  # count rows in each group
specialty_by_age = specialty_by_age.reset_index(name="count")  # turn counts into a table

age_specialty_totals = specialty_by_age.groupby("age_group")["count"].transform("sum")  # total for each age group

specialty_by_age["percentage_within_age_group"] = specialty_by_age["count"] / age_specialty_totals  # make a share
specialty_by_age["percentage_within_age_group"] = specialty_by_age["percentage_within_age_group"] * 100  # make it percent
specialty_by_age["percentage_within_age_group"] = specialty_by_age["percentage_within_age_group"].round(1)  # round it 

specialty_by_age = specialty_by_age.sort_values(["age_group", "count"], ascending=[True, False])  # sort biggest first

print("Specialty by age group")  # label the table
print(specialty_by_age)  # show the table
```
